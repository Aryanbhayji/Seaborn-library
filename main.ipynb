```python
data= sns.load_dataset('tips')
data
```

```python
'''Seaborn Plot
     │
     ├── data → What data?
     ├── x    → X-axis?
     ├── y    → Y-axis?
     ├── hue  → Color groups?
     ├── size → Size groups?
     ├── style → Marker/style groups?
     └── palette → Which colors?'''
sns.scatterplot(data,x="total_bill",y="tip" , hue="sex",size="size")
```

```python
#plot pairwise relationships in a dataset

penguins=sns.load_dataset("penguins").dropna()
print(penguins)
sns.pairplot(penguins)
```

```python
plt.figure(figsize=(6,4))
sns.pairplot(penguins , hue="species" , markers=["o","s","D"], kind='scatter')
sns.set_theme(style="darkgrid")
```

```python
penguins = sns.load_dataset("penguins")
g = sns.PairGrid(penguins,hue="species")
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
g.map_diag(sns.kdeplot ,color=".3")
sns.set_palette(palette="dark:#5A9_r")
```

```python
x_vars = ["body_mass_g", "bill_length_mm", "bill_depth_mm", "flipper_length_mm"]
y_vars = ["body_mass_g"]
g = sns.PairGrid(penguins, hue="species", x_vars=x_vars, y_vars=y_vars)
g.map_diag(sns.histplot, color=".3")
g.map_offdiag(sns.scatterplot)
g.add_legend()
```
