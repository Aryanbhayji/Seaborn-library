```python
data= sns.load_dataset('tips')
data
```

```python
'''Seaborn Plot
     │
     ├── data → What data?
     ├── x    → X-axis?
     ├── y    → Y-axis?
     ├── hue  → Color groups?
     ├── size → Size groups?
     ├── style → Marker/style groups?
     └── palette → Which colors?'''
sns.scatterplot(data,x="total_bill",y="tip" , hue="sex",size="size")
```

```python
sns.relplot(data,x="total_bill",y="tip" , hue="sex",size="size",kind="scatter")
```

```python
sns.relplot(data,x="total_bill",y="tip" , hue="sex",size="size",col="day",col_wrap=2,kind="scatter")
```

```python
sns.displot(data,x="tip",col="sex",kind="hist")
```

```python
sns.histplot(data,x="tip",hue="sex",bins=10)
```

```python
sns.kdeplot(data=data,x="tip",y="total_bill",hue="sex")
```

```python
#Plot pairwise relationships in a dataset.
sns.pairplot(data=data,hue="sex")
sns.set_theme(style="darkgrid")
```

```python
#Plot data and a linear regression model fit.
sns.regplot(data=data,x="total_bill",y="tip",scatter=True)
```

```python
#Plot data and regression model fits across a FacetGrid.
sns.lmplot(data=data,x="total_bill",y="tip")
```

```python
#Draw a box plot to show distributions with respect to categories
#Detect outliers + compare distributions
car_data=sns.load_dataset("mpg").dropna()
cars_data=car_data[car_data.cylinders.isin([4,6,8])]
sns.boxplot(x=cars_data.mpg)
```

```python
cars_data.mpg.describe()
```

```python
sns.boxplot(x=cars_data.origin,y=cars_data.mpg)
cars_data.groupby("origin").mpg.describe()
```
