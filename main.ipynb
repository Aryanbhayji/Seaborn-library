```python
data= sns.load_dataset('tips')
data
```

```python

sns.scatterplot(data,x="total_bill",y="tip" , hue="sex",size="size")
```

```python
sns.relplot(data,x="total_bill",y="tip" , hue="sex",size="size",kind="scatter")
```

```python
sns.relplot(data,x="total_bill",y="tip" , hue="sex",size="size",col="day",col_wrap=2,kind="scatter")
```
